```python
import pandas as pd

INPUT_FILE = "wcer (1) (1) (1).csv"
OUTPUT_FILE = "wcer_preprocessed.csv"
REQUIRED_COLUMNS = ["STORE", "UPC", "WEEK", "MOVE", "PRICE"]

def main():
    # Make sure your file is uploaded and name matches INPUT_FILE
    df = pd.read_csv(INPUT_FILE)
    df.columns = df.columns.str.strip().str.upper()

    missing_columns = [col for col in REQUIRED_COLUMNS if col not in df.columns]

    if missing_columns:
        raise ValueError(f"Missing columns: {missing_columns}")

    df = df[REQUIRED_COLUMNS].copy()

    for col in REQUIRED_COLUMNS:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    print("Total rows:", len(df))
    print("\nMissing or invalid values:")
    print(df.isna().sum())

    print("\nExact duplicate rows:", df.duplicated().sum())

    print("\nNegative values to review:")
    for col in REQUIRED_COLUMNS:
        print(f"{col}: {(df[col] < 0).sum()}")

    repeated = df.duplicated(subset=["STORE", "UPC", "WEEK"], keep=False).sum()
    print("\nRepeated store/product/week rows:", repeated)

    before = len(df)
    df = df.dropna(subset=REQUIRED_COLUMNS)
    print("\nRows removed:", before - len(df))

    df.to_csv(OUTPUT_FILE, index=False)
    print("\nSaved:", OUTPUT_FILE)

# Run the function immediately
if __name__ == "__main__":
    main()
```
